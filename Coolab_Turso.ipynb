```python
import libsql

# Tus credenciales de Turso
url = "libsql://sistema-de-inventario-herson-06.aws-us-west-2.turso.io"
auth_token = "eyJhbGciOiJFZERTQSIsInR5cCI6IkpXVCJ9.eyJhIjoicnciLCJpYXQiOjE3ODg4Nzg0MjQsImlkIjoiMDFhMDgxNjAtNDcwMS03NTgzLWE0YjktZjIzN2I5ZWRlN2E4Iiwia2lkIjoiU1VrbUM1VENVQjJidWhIaFJWajFfWjZYOGtKNWxJek9BQlhkODlsam5wNCIsInJpZCI6IjI4OTU3ODE1LWViMjYtNGE4NC04NmY2LWM0OTU2OWUwNGQzZiJ9.bQidQ_ZDO8ViIlQtZVpw4_IRf3wI2vNXGcSl-AhmSIu9GrRIiWph_p0SdCcf6DZjkRifFNq7W72Nm4loqwBoCw"

# Conectar y sincronizar con Turso
conn = libsql.connect("inventario_local.db", sync_url=url, auth_token=auth_token)
conn.sync()
cursor = conn.cursor()

# ==========================================
# ENTRADA MANUAL INTERACTIVA
# ==========================================
print("--- REGISTRO DE NUEVA CATEGORÍA ---")
nombre_categoria = input("Ingresa el nombre de la categoría: ")
descripcion_categoria = input("Ingresa la descripción de la categoría: ")

# 1. Insertar el registro usando los datos que escribiste
cursor.execute(
    "INSERT INTO categorias (nombre, descripcion) VALUES (?, ?);",
    (nombre_categoria, descripcion_categoria)
)

# 2. Guardar localmente y sincronizar con la nube de Turso
conn.commit()
conn.sync()

print("\n¡Datos guardados y sincronizados con éxito en Turso!\n")

# ==========================================
# 3. MOSTRAR TODAS LAS CATEGORÍAS REGISTRADAS
# ==========================================
cursor.execute("SELECT * FROM categorias;")
registros = cursor.fetchall()

print("Listado actual de la tabla 'categorias':")
for reg in registros:
    print(f"ID: {reg[0]} | Nombre: {reg[1]} | Descripción: {reg[2]}")

# Cerrar conexión
conn.close()
```
